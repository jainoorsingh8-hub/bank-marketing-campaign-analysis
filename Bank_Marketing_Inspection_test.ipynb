# Bank Marketing Data Analysis
This notebook presents an analysis of the `bankmarketing.csv` dataset, which includes data related to a bank's marketing campaigns. The main goal is to understand customer behavior and predict whether a client will subscribe to a term deposit.

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv("bankmarketing.csv")

# Display the first 5 rows
df.head()

# Display number of rows and columns
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())

# Display basic information
print("\nDataset Information:")
df.info()
```

```python

# Check for missing values
print("Missing Values in Each Column:")
print(df.isnull().sum())

# Check percentage of missing values
print("\nPercentage of Missing Values:")
print((df.isnull().sum() / len(df) * 100).round(2))

# Check duplicate rows
print("\nNumber of Duplicate Rows:", df.duplicated().sum())

# Display data types
print("\nData Types:")
print(df.dtypes)
```

```python

# Summary statistics for numerical columns
print("Numerical Summary Statistics:")
display(df.describe())

# Summary statistics for categorical columns
print("\nCategorical Summary Statistics:")
display(df.describe(include="object"))
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

# Plot the distribution of the target variable

# Count target variable values
target_counts = df["y"].value_counts()

print("Target Variable Distribution:")
print(target_counts)

# Plot target distribution
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="y")

plt.title("Distribution of Term Deposit Subscription")
plt.xlabel("Subscribed to Term Deposit")
plt.ylabel("Number of Customers")

plt.show()
```

```python
# Calculate percentage distribution
target_percentage = df["y"].value_counts(normalize=True) * 100

print("Target Variable Percentage:")
print(target_percentage.round(2))

# Pie chart
plt.figure(figsize=(6, 6))

plt.pie(
    target_counts,
    labels=target_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Term Deposit Subscription Percentage")

plt.show()
```

```python
# Select numerical columns
numerical_columns = df.select_dtypes(include=np.number).columns

# Calculate correlation matrix
correlation_matrix = df[numerical_columns].corr()

# Plot correlation matrix
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Matrix of Numerical Features")
plt.show()
```

```python
# Plot distributions of numerical columns
numerical_columns = df.select_dtypes(include=np.number).columns

for column in numerical_columns:
    plt.figure(figsize=(7, 4))
    
    sns.histplot(
        data=df,
        x=column,
        kde=True
    )
    
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    
    plt.show()
```

```python
# Subscription rate by job
job_subscription = pd.crosstab(
    df["job"],
    df["y"],
    normalize="index"
) * 100

print(job_subscription.round(2))

# Plot
plt.figure(figsize=(12, 6))

job_subscription["yes"].sort_values().plot(
    kind="bar"
)

plt.title("Term Deposit Subscription Rate by Job")
plt.xlabel("Job")
plt.ylabel("Subscription Rate (%)")
plt.xticks(rotation=45)

plt.show()
```

```python

# Subscription by education
education_subscription = pd.crosstab(
    df["education"],
    df["y"],
    normalize="index"
) * 100

print(education_subscription.round(2))

plt.figure(figsize=(10, 5))

education_subscription["yes"].sort_values().plot(
    kind="bar"
)

plt.title("Term Deposit Subscription Rate by Education")
plt.xlabel("Education")
plt.ylabel("Subscription Rate (%)")
plt.xticks(rotation=45)

plt.show()
```

```python
# Subscription by martial status
marital_subscription = pd.crosstab(
    df["marital"],
    df["y"],
    normalize="index"
) * 100

print(marital_subscription.round(2))

plt.figure(figsize=(8, 5))

marital_subscription["yes"].plot(
    kind="bar"
)

plt.title("Term Deposit Subscription Rate by Marital Status")
plt.xlabel("Marital Status")
plt.ylabel("Subscription Rate (%)")

plt.show()
```

```python
# Age analysis
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="y",
    y="age"
)

plt.title("Age Distribution by Term Deposit Subscription")
plt.xlabel("Subscribed")
plt.ylabel("Age")

plt.show()
```

```python
# Campaign analysis
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="y",
    y="campaign"
)

plt.title("Number of Contacts During Campaign by Subscription")
plt.xlabel("Subscribed")
plt.ylabel("Number of Contacts")

plt.show()
```

```python
#monthly wise subscription analysis
month_subscription = pd.crosstab(
    df["month"],
    df["y"],
    normalize="index"
) * 100

print(month_subscription.round(2))

plt.figure(figsize=(10, 5))

month_subscription["yes"].plot(
    kind="bar"
)

plt.title("Term Deposit Subscription Rate by Month")
plt.xlabel("Month")
plt.ylabel("Subscription Rate (%)")

plt.show()
```

```python
# Previous campaign outcome
poutcome_subscription = pd.crosstab(
    df["poutcome"],
    df["y"],
    normalize="index"
) * 100

print(poutcome_subscription.round(2))

plt.figure(figsize=(8, 5))

poutcome_subscription["yes"].plot(
    kind="bar"
)

plt.title("Subscription Rate by Previous Campaign Outcome")
plt.xlabel("Previous Campaign Outcome")
plt.ylabel("Subscription Rate (%)")

plt.show()
```

```python

print("===== KEY INSIGHTS =====")

# Overall subscription rate
subscription_rate = (df["y"] == "yes").mean() * 100
print(f"\n1. Overall subscription rate: {subscription_rate:.2f}%")

# Most common job
most_common_job = df["job"].mode()[0]
print(f"2. Most common job category: {most_common_job}")

# Average age
print(f"3. Average customer age: {df['age'].mean():.2f} years")

# Average call duration
print(f"4. Average call duration: {df['duration'].mean():.2f} seconds")

# Most successful previous campaign outcome
poutcome_rate = pd.crosstab(
    df["poutcome"],
    df["y"],
    normalize="index"
) * 100

best_poutcome = poutcome_rate["yes"].idxmax()

print(f"5. Previous campaign outcome with highest subscription rate: {best_poutcome}")

# Job with highest subscription rate
job_rate = pd.crosstab(
    df["job"],
    df["y"],
    normalize="index"
) * 100

best_job = job_rate["yes"].idxmax()

print(f"6. Job category with highest subscription rate: {best_job}")
```


## Conclusion

The Bank Marketing dataset contains 41,188 customer records and 21 features describing customer demographics, financial information, previous campaign outcomes, and current marketing campaign details.

The exploratory data analysis shows that the target variable `y` is imbalanced, with considerably more customers not subscribing to a term deposit than customers who subscribe. This is an important consideration for future machine learning modeling.

The analysis also shows that customer characteristics such as job, education, marital status, previous campaign outcome, campaign contacts, and call duration can provide useful insights into subscription behavior.

The correlation analysis helps identify relationships among numerical variables, while categorical analysis highlights groups with different subscription rates.

Overall, this exploratory analysis provides a strong foundation for the next stages of the project, including data preprocessing, categorical encoding, feature selection, handling class imbalance, and building machine learning classification models to predict whether a customer will subscribe to a term deposit.
